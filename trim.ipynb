```python
#write for loop that takes every column besides date from csv file and appends column to bothdata.csv

import csv
import os
import numpy as np
from datetime import datetime, timedelta

#should have 1044 columns
#write function that will read in the first 1044 columns of data from a csv file with year starting in 2004 (this uses dates). This should work for any csv file with the same format and adds columns to numpy array data
def read_data(filename,data):
    with open(filename, 'r') as file:
        reader = csv.reader(file)
        next(reader)  # Skip the header row
        # make first date in each month the first day of the month
        current_date = datetime(2004, 1, 4)
        for idx, row in enumerate(reader):
            # Convert all values in the row to float except for the first column
            #combine all these values in one column
            #if current date has day number 1-6, then set current date to the first day of the month
            temp_curr_date = current_date
            if current_date.day <= 7:
                temp_curr_date = datetime(current_date.year, current_date.month, 1)
            row_data = [float(val) for val in row[1:]]
            if temp_curr_date not in data:
                data[temp_curr_date] = []
            data[temp_curr_date].append(row_data)
            current_date += timedelta(days=7)
            #if year of datetime is 2024, then break
            if current_date.year == 2024:
                break

def read_data_monthly(filename,data):
    with open(filename,'r') as file:
        reader = csv.reader(file)
        next(reader)
        #make sure DATE col is more than or equal to 2004, 1, 1, otherwise filter
        for row in reader:
            current_date = datetime.strptime(row[0], '%Y-%m-%d')
            if current_date >= datetime(2003,12,1):
                break
        #how to go back one element in reader
        
        for idx, row in enumerate(reader):
            current_date = datetime.strptime(row[0], '%Y-%m-%d')
            if current_date >= datetime(2024,1,1):
                break
            row_data = [float(val) for val in row[1:]]
            if current_date not in data:
                data[current_date] = []
            data[current_date].append(row_data)



        
def main():
    folder_path = "/Users/kenjiwagner/Desktop/eecs448/projectstuff/csvfiles"  # Change this to the path of your folder
    data = {}

    # Iterate over all files in the folder
    for filename in os.listdir(folder_path):
        if filename.endswith(".csv"):
            file_path = os.path.join(folder_path, filename)
            read_data(file_path, data)
            
    print(len(data))
    with open("/Users/kenjiwagner/Desktop/eecs448/projectstuff/macro1.csv" , 'w', newline='') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(['start', 'ffrate', 'gasregcovw', 'gasregw', 'insured_unemployment', 'nfcilevrage','stlifsi4'])  # Write header
        for date, rows in data.items():
            writer.writerow([date.strftime('%Y-%m-%d'), rows[0][0], rows[1][0], rows[2][0], rows[3][0], rows[4][0], rows[5][0]])  # Write data
    folder_path2 = "/Users/kenjiwagner/Desktop/eecs448/projectstuff/monthlycsvfiles"
    data2 = {}
    for filename in os.listdir(folder_path2):
        if filename.endswith(".csv"):
            file_path = os.path.join(folder_path2, filename)
            read_data_monthly(file_path, data2)
    print(len(data2))
    with open("/Users/kenjiwagner/Desktop/eecs448/projectstuff/macro_monthly.csv" , 'w', newline='') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(['start', "totalhealthcareconstrspend", "oneyearrealinterestrate", 'twoyearexpectedinflationrate',"inventorytosalesratio",'capacityutiltotlidx', 'apgradeaeggs','totalvehiclesales',"inflationexpectation"])  # Write header
        for date, rows in data2.items():
            writer.writerow([date.strftime('%Y-%m-%d')] + [row[0] if row else '' for row in rows[:9]])
```
